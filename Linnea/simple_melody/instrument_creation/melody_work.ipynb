```python
import librosa
import IPython.display as ipd
import numpy as np
import soundfile as sf
```

```python
def generate_adsr_envelope(duration, envelope_values, start=0, sr=44100):
    if np.array(envelope_values).shape != (3, 2): # amp value + percentage for A, D, & S (R = 0 at 100% duration)
        raise ValueError("generate_adsr_envelope accepts an amplitude value and percentage for attack, decay, and sustain (release = 0 at 100% duration)")
    if len([l for (a, l) in envelope_values if l < 0 or l > 1]) != 0:
        raise ValueError("invalid lengths for envelope; lengths should be a percentage of the duration")
    lengths = [l for (a, l) in envelope_values]
    if not np.array_equal(lengths, np.sort(lengths)):
        raise ValueError("ADSR values must be in order")

    total_samples = 0
    prev_amp = start
    segments = []
    for (amp, length) in envelope_values:
        cur_samples = int(np.ceil(sr * duration * length)) - total_samples
        segments.append(np.linspace(prev_amp, amp, num=cur_samples))
        total_samples += cur_samples
        prev_amp = amp

    # release
    release_samples = int(np.ceil(sr * duration)) - total_samples
    segments.append(np.linspace(prev_amp, 0, num=release_samples))
    
    t = np.concatenate(segments)

    return t

# adsr = [(1, 0), (0.8, 0.4), (0.8, 0.8)]
# env = generate_adsr_envelope(1, adsr)
```

```python
# marimba
filename = '../sound_library/combos_linnea/fullex_rev_rand.wav'
fullex_rev_rand, srate = librosa.load(filename, sr=None)

adsr = [(1, 0), (0.8, 0.4), (0.8, 0.8)]
mallet_env = generate_adsr_envelope(len(fullex_rev_rand)/44100, adsr)

ipd.Audio(fullex_rev_rand * mallet_env, rate=srate)
sf.write("marimba.wav", fullex_rev_rand * mallet_env / np.max(np.abs(fullex_rev_rand * mallet_env)), srate)
```
