# Automatic Orchestration
Automatic orchestration involves automatically assigning instruments, sounds, articulations, dynamics to the notes of an unorchestrated score or even changing the score itself (Handelman et al. pp. 43, 2012).

In "Live Orchestral Piano, a System for Real-Time Orchestral Music Generation" (2016-2017), Crestel and Esling use a machine-learning model called a Conditional Restricted Boltzmann Machine (cRBM) to automatically generate orchestral arrangements from piano performances. Their system is called Live Orchestral Piano and they have published a github repository of their method: https://github.com/qsdfo/LOP. The main idea is that instead of manually programming orchestration rules like "assign higher notes to the flute and lower notes to the bass", they train a neural network to learn how composers arrange piano music for orchestras.

In *"Research on Piano Harmony Automatic Orchestration System Based on Deep Learning"* by Yun Cui and Meng Qin (2024), the authors describe a system that uses deep learning techniques to extract musical information including notes, timing, and dynamics, and automatically generate musical scores. They do this by training neural networks on various piano melodies, then convert all the notes in a piano score to vectors, then use a system of automatic arrangement based on the deep learning (Cui & Qin pp. 59, 2024).

"The process of computer music arrangement is to find a suitable set of chord accompaniment for the whole melody by using the development trend of the main melody and the characteristic state of each melody through the computer algorithm." (Cui & Qin pp. 59-60, 2024).

The paper mainly focuses on automatic harmony arrangement for piano melodies, rather than the broader task of orchestration which involves distributing music across different instruments (Cui & Qin pp. 59-59, 2024). However, it is a good source to explain harmonic accompaniment. In the paper, the authors explain that harmony orchestration involves adding harmonies to the main melody based on the relationships between successive notes (Cui & Qin pp. 62, 2024). They distinguish between vertical harmony (simultaneous notes) and horizontal harmony (harmonic progression over time) (Cui & Qin pp. 62, 2024)

## Could we use Automatic Orchestration on Toms Diner?
Yes, but it will require a lot more work than most technique exploration, hense I think we might want to explore this as a potential direction for part 2 of the project, rather than initial technique exploration.

<br/>

For our possible implementation, we would need to do the following
1. Get a MIDI score of Tom's Diner
2. Find suitable orchestration training data. -- We could use the paired piano and orchestral MIDI dataset provided by Crestel and Esling or investigate other available datasets.
3. Choose an orchestration model. -- Could adapt an approach similar to cRBM, adapt the python implementation by Crestel and Esling, or investigate newer models
4. Train or adapt the model
5. Generate and evaluate an orchestration -- Using the MIDI representation of Tom's Diner as input and generate an arrangement containing multiple instrumental parts, then synthesize it into audio and compare different arrangements.
<br/>

Since most of the research on automatic orchestration we've looked at was from a long time ago, there may be newer AI-based methods that could simplify the implementation. It would be interesting to investigate these approaches without needing to develop a complex neural network from scratch. However, this would still be experimental therefore the quality of the generated orchestration could vary depending on a lot of different factors.
