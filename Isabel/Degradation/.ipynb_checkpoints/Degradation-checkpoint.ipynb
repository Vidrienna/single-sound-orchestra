# Degradation
Audio degradation is a general term for the decrease in perceived quality of sound over time.

William Basinski’s The Disintegration Loops (https://williambasinski.bandcamp.com/album/the-disintegration-loops) is an example of analog degradation using magnetic tape.

There have been attempts to emulate analog degradation digitally, most notably in the Matlab Degradation Toolbox and its Python implementation (see References). There is not a single technique that is used to degrade audio; rather, it is a combination of several methods of manipulating audio.
